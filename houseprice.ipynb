```python
import sys
class Employee:
    def show(self):
        print(f"the python version is {sys.version}")
    def __init__(self,name:str,address:str,phone:int): # ITS NOW A CONSTRUCTOR
        print('----I AM A CONSTRUCTOR---')
        self.name=name
        self.address=address
        self.phone=phone
    def showemp(self):
        print(f"name is {self.name}, address is {self.address} and phone is {self.phone}.")

e1=Employee("Uttaran","Kolkata",1234)
e1.showemp()
        
```

```python
import sys
class Employee:
    def show(self):
        print(f"the python version is {sys.version}")
    def __init__(self,name:str,address:str,phone:int): # ITS NOW A CONSTRUCTOR
        print('----I AM A CONSTRUCTOR---')
        self.name=name
        self.address=address
        self.phone=phone
    def __str__(self):
        return f"name is {self.name}, address is {self.address} and phone is {self.phone}."

e1=Employee("Uttaran","Kolkata",1234)
print(e1.__str__())
print(e1)
```

```python
def brain(model:LinearRegression,sqft:int):
    return model.coef_ * sqft + model.intercept_
```
